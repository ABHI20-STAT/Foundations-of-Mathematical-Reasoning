#🧩1.**Proposition**

##**Theoretical Background:**

A **proposition** is any declarative sentence that has a **definite truth value** — either **True (T)** or **False (F)**.

In logic, propositions can be:

- **Atomic (simple)**: Cannot be broken further (e.g., “2 + 2 = 4”)
- **Compound**: Formed by joining atomic propositions with logical connectives such as:
  - **NOT** (¬)
  - **AND** (∧)
  - **OR** (∨)
  - **XOR** (⊕)
  - **IMPLIES** (→)

##**Example:**

Let:  
- \( P \): “It is raining.”  
- \( Q \): “The ground is wet.”

Then the compound proposition is:  

$$ P \rightarrow Q $$

**Interpretation**: “If it is raining, then the ground is wet.”


```r
# Logical propositions
P <- c(TRUE, TRUE, FALSE, FALSE)
Q <- c(TRUE, FALSE, TRUE, FALSE)

# Implication: P → Q is equivalent to (!P) | Q
Implication <- (!P) | Q

# Combine in a data frame
data.frame(P, Q, Implication)
```

This is a proposition — specifically a **contingency**, since it’s sometimes true, sometimes false.

#⚖️**Tautology**

##**Theoretical Definition:**  

A tautology is a compound statement that is always true, no matter the truth values of its components.

$$P∨¬P$$

is the **Law of Excluded Middle** — a fundamental tautology in classical logic.

```r
P <- c(TRUE, FALSE)

# Law of Excluded Middle: P ∨ ¬P
Tautology <- P | (!P)

data.frame(P, not_P = !P, Tautology)
```

✅ `all(Tautology)` returns **TRUE** → proves it’s a tautology.

`all(Tautology)  # Checks if all results are TRUE`

## **Interpretation:**

Even if $𝑃$ is true or false, $\textbf{P or not P}$ must always be true. This is one of the logical laws used in every formal proof system.


#💥**3. Contradiction**
## **Theoretical Definition:**

A contradiction is a compound statement that is always false, regardless of the truth values of its components.

$$𝑃∧¬𝑃$$

is the **Law of Non-Contradiction** — it can never be true.

```r
P <- c(TRUE, FALSE)

# Law of Non-Contradiction: P ∧ ¬P
Contradiction <- P & (!P)

data.frame(P, not_P = !P, Contradiction)

```

❌ `any(Contradiction)` returns **FALSE** → proves it’s a contradiction.

`any(Contradiction)  # Checks if any TRUE values exist`

## **Interpretation:**

You cannot have $𝑃$ and $¬𝑃$ true at the same time — this violates the principle of consistency.

# **🧠4.Contingency**

## **Theoretical Definition:**

A contingency is a compound proposition that is sometimes true and sometimes false, depending on the truth values of its components.

Example:

$$P→Q$$



```r
P <- c(TRUE, TRUE, FALSE, FALSE)
Q <- c(TRUE, FALSE, TRUE, FALSE)

# Implication: P → Q = !P | Q
Contingency <- (!P) | Q

data.frame(P, Q, Contingency)
```

⚙️ `unique(Contingency)` returns both `TRUE` and `FALSE` → it’s a contingency.

## 📘 Summary Table: Types of Logical Statements

| **Type**        | **Logical Form** | **Always True?** | **Always False?** | **Example**                    |
|-----------------|------------------|------------------|-------------------|--------------------------------|
| Proposition     | $P$          | ❌               | ❌                | "2 + 2 = 4"                    |
| Tautology       | $P \lor \neg P$ | ✅            | ❌                | Law of Excluded Middle        |
| Contradiction   | $ P \land \neg P $ | ❌           | ✅                | Law of Non-Contradiction      |
| Contingency     | $ P \rightarrow Q $ | ❌          | ❌                | Conditional statement         |

# **🧮 Bonus: Automatic Detection in R**

Here’s a reusable R function that classifies any logical statement:

```r
classify_logic <- function(x) {
  if (all(x)) {
    return("Tautology ✅")
  } else if (!any(x)) {
    return("Contradiction ❌")
  } else {
    return("Contingency ⚙️")
  }
}

# Example
P <- c(TRUE, FALSE)
Q <- c(TRUE, FALSE)
expr <- (!P) | Q  # Implication form
classify_logic(expr)
```
